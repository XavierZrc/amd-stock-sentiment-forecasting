```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, accuracy_score

from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB
```

```python
df_title = pd.read_excel("FinBERT_Title_Only_Labeled.xlsx")
df_full  = pd.read_excel("FinBERT_Title_and_Post_Labeled.xlsx")
```

```python
# Title_Only
df_title = df_title.dropna(subset=["title", "finbert_label"]).copy()

X_text_title = df_title["title"]
y_title = df_title["finbert_label"]
```

```python
le_title = LabelEncoder()
y_title_enc = le_title.fit_transform(y_title)
```

```python
tfidf_title = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    stop_words="english"
)

X_title_tfidf = tfidf_title.fit_transform(X_text_title)
```

```python
Xtr_t, Xte_t, ytr_t, yte_t = train_test_split(
    X_title_tfidf,
    y_title_enc,
    test_size=0.2,
    random_state=42,
    stratify=y_title_enc
)
```

```python
# SVM title
svm_title = SVC(kernel="linear", probability=True, random_state=42)
svm_title.fit(Xtr_t, ytr_t)

pred_svm_title = svm_title.predict(Xte_t)

print("SVM — Title Only")
print("Accuracy:", accuracy_score(yte_t, pred_svm_title))
print(classification_report(yte_t, pred_svm_title, target_names=le_title.classes_))
```

```python
nb_title = MultinomialNB()
nb_title.fit(Xtr_t, ytr_t)

pred_nb_title = nb_title.predict(Xte_t)

print("Naive Bayes — Title Only")
print("Accuracy:", accuracy_score(yte_t, pred_nb_title))
print(classification_report(yte_t, pred_nb_title, target_names=le_title.classes_))
```

```python
# Title+post
df_full = df_full.dropna(subset=["full_text", "finbert_label"]).copy()

X_text_full = df_full["full_text"]
y_full = df_full["finbert_label"]
```

```python
le_full = LabelEncoder()
y_full_enc = le_full.fit_transform(y_full)
```

```python
tfidf_full = TfidfVectorizer(
    max_features=8000,
    ngram_range=(1, 2),
    stop_words="english"
)

X_full_tfidf = tfidf_full.fit_transform(X_text_full)
```

```python
Xtr_f, Xte_f, ytr_f, yte_f = train_test_split(
    X_full_tfidf,
    y_full_enc,
    test_size=0.2,
    random_state=42,
    stratify=y_full_enc
)
```

```python
svm_full = SVC(kernel="linear", probability=True, random_state=42)
svm_full.fit(Xtr_f, ytr_f)

pred_svm_full = svm_full.predict(Xte_f)

print("SVM — Title + Post")
print("Accuracy:", accuracy_score(yte_f, pred_svm_full))
print(classification_report(yte_f, pred_svm_full, target_names=le_full.classes_))
```

```python
nb_full = MultinomialNB()
nb_full.fit(Xtr_f, ytr_f)

pred_nb_full = nb_full.predict(Xte_f)

print("Naive Bayes — Title + Post")
print("Accuracy:", accuracy_score(yte_f, pred_nb_full))
print(classification_report(yte_f, pred_nb_full, target_names=le_full.classes_))
```
